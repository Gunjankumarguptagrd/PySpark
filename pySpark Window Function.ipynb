```python
#initiate spark
import pyspark
from pyspark import SparkContext, SparkConf
from pyspark.sql import SparkSession
c = pyspark.SparkConf().setAppName("test_app").setMaster("local")
sc = pyspark.SparkContext(conf = c)
spark = SparkSession(sc)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window
from pyspark.sql.functions import desc, rank
```

```python
data = spark.read.csv(r"D:\Data Engineering\All Dataset\student_info.csv", header = True, inferSchema = True)
```

```python
data.show()
```

```python
data = data.withColumn('Rank1', rank().over(Window.orderBy(desc("MARKS"))))
```

```python
data = data.withColumn('D_Rank1', dense_rank().over(Window.orderBy(desc("MARKS"))))
```

```python
data.show()
```
