```python
#initiate spark
import pyspark
from pyspark import SparkContext, SparkConf
from pyspark.sql import SparkSession
c = pyspark.SparkConf().setAppName("test_app").setMaster("local")
sc = pyspark.SparkContext(conf = c)
spark = SparkSession(sc)
```

```python
data = spark.read.csv("D:\Data Engineering\Data Set\Car_data.csv",inferSchema=None, header = True)
```

```python
data.show(5)
```

```python
data.printSchema()
```

```python
#data[data['Year']==2014]["Car_Name"].count() ------ pandas
data[data['Year'] == 2014].count()
```

```python
data.select('Car_Name','Year').show(5)
```

```python
from pyspark.sql.functions import *
```

```python
data.select('Car_Name','Year',"Fuel_Type").filter(col("Fuel_Type")=='Petrol').show(5)
```

```python
data.groupby('Year').count().sort(desc('count')).show(5)
```

```python
data.createOrReplaceTempView("car_data")
```

```python
spark.sql("select * from car_data").show(5)
```

```python
spark.sql("select Car_Name, Year from car_data where Fuel_Type = 'Petrol'").show(5)
```

```python
spark.sql("select Year,count(*) total_cars from car_data group by year having count(*)>30").show()
```
