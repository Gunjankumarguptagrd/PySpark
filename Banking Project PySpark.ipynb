```python
def check_missing_value(df,lst_cl):  # lst_cl = ['id', 'merber_id','loan_amt'] ect % will get
    missing_values_percent = {}
    for i in lst_cl:
        a = df.filter(col(i).isNull()).count()
        b = df.count()
        c = (a/b) * 100
        missing_values_percent[i] = c
    return missing_values_percent
```

```python
def check_missing_percent(df,lst_cl):
    global missing_values_less_75
    global missing_values_more_75
    missing_values_less_75 = {}
    missing_values_more_75 = {}
    b = df.count()
    for i in lst_cl:
        a = df.filter(col(i).isNull()).count()
        c = (a/b) * 100
        if c >= 75:
            missing_values_more_75[i] = c
        else:
            missing_values_less_75[i] = c
    return({"missing_values_more than 75% " :missing_values_more_75, "missing_values_less than 75%" :missing_values_less_75})
    
```

```python
# to check duplicate values
def check_duplicates(df,cl):
    a = df.select(cl).distinct().count()
    b = df.count()
    if a==b:
        return "No Duplicate"
    else:
        c = b-a
        Print("There are ",c,"Duplicate")
```
